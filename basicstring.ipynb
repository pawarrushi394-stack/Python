```python
s = "ADOBECODEBANC"
pattern = "ABC"

min_window = ""

for i in range(len(s)):
    for j in range(i + 1, len(s) + 1):

        window = s[i:j]

        if all(ch in window for ch in pattern):

            if min_window == "" or len(window) < len(min_window):
                min_window = window

print("Minimum window =", min_window)
```

```python
s = input("Enter string: ")

for i in range(1, len(s) + 1):

    if len(s) % i == 0:
        pattern = s[:i]
        count = len(s) // i

        if pattern * count == s:
            print("Repeated pattern =", pattern)
            print("Repeat count =", count)
            break
```

```python
words = ['eat', 'tea', 'tan', 'ate', 'nat', 'bat']

groups = {}

for word in words:
    key = ''.join(sorted(word))

    if key not in groups:
        groups[key] = []

    groups[key].append(word)

print(list(groups.values()))
```

```python
words = ['flower', 'flow', 'flight']

prefix = words[0]

for word in words:
    while not word.startswith(prefix):
        prefix = prefix[:-1]

print("Longest common prefix =", prefix)
```

```python
a = "{[()()]}"
stack = []

for ch in a:

    if ch in "{([":
        stack.append(ch)

    elif ch in "})]":

        if not stack:
            print("Balanced = False")
            break

        if (ch == ")" and stack[-1] != "(") or \
           (ch == "}" and stack[-1] != "{") or \
           (ch == "]" and stack[-1] != "["):
            print("Balanced = False")
            break

        stack.pop()

else:
    print("Balanced =", len(stack) == 0)
        
        
    
```

```python
# 4.  Write a Python program to decode a run-length encoded string.
# Input a3b2c4
# Expected Output Decoded string = aaabbcccc
a = "a3b2c4"

b = ""
i = 0

while i < len(a):
    char = a[i]
    count = int(a[i + 1])

    b += char * count

    i += 2
print(b)


```
