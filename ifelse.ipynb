```python
# 17. Write a Python program to input electricity unit charges and calculate total 
# electricity bill according to the given condition:
# For first 50 units Rs. 0.50/unit
# For next 100 units Rs. 0.75/unit
# For next 100 units Rs. 1.25/unit
# For unit above 250 Rs. 1.50/unit
# An additional surcharge of 17% is added to the bill

units = float(input("Enter electricity units: "))

if units <= 50:
    bill = units * 0.50

elif units <= 150:
    bill = (50 * 0.50) + ((units - 50) * 0.75)

elif units <= 250:
    bill = (50 * 0.50) + (100 * 0.75) + ((units - 150) * 1.25)

else:
    bill = (50 * 0.50) + (100 * 0.75) + (100 * 1.25) + ((units - 250) * 1.50)

surcharge = bill * 17 / 100
total_bill = bill + surcharge

print("Electricity bill =", bill)
print("Surcharge =", surcharge)
print("Total bill =", total_bill)
```

```python
# 16. Write a Python program to input basiPython salary of an employee and 
# calculate its Gross salary according to following:
# Basic Salary <= 10000 : HRA = 20%, DA = 80%
# Basic Salary <= 20000 : HRA = 30%, DA = 90%
# Basic Salary > 20000 : HRA = 35%, DA = 95%

salary = int(input("Enter a salary:"))
if salary <= 10000:
    hra = 20/100 * salary
    da = 80/100 * salary
elif  salary <= 20000:
    hra = 30/100 * salary
    da = 90/100 * salary
else:
    hra = 35/100 *salary
    da = 95/100 * salary

gross = salary + hra + da

print("Gross Salary :",gross)
```
