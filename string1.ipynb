```python
name = str(input("Enter the string:"))
count_uppercase = 0
count_lowercase = 0
count_digits = 0
count_special = 0
for i in name:
    if i.isupper():
        count_uppercase += 1
    elif i.islower():
        count_lowercase += 1
    elif i.isdigit():
        count_digits += 1
    else:
        count_special += 1
print("Number of uppercase letters:",count_uppercase)
print("Number of lowercase letters:",count_lowercase)
print("Number of digits:",count_digits)
print("Number of special characters:",count_special)
```

```python
name = str(input("Enter the string to know number of vowels and consonants"))
count_vowels = 0
count_consonants = 0
for i in name:
    if i in 'aeiouAEIOU':
        count_vowels += 1
    else:
        count_consonants += 1
print("Number of vowels:",count_vowels)
print("Number of consonants:",count_consonants)
```
